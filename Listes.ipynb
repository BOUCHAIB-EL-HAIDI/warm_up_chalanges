<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 1 : La liste des notes </span></H3>

<center><code>notes = [12, 4, 14, 11, 18, 13, 7, 10, 5, 9, 15, 8, 14, 16]</code></center>

<span style="font-weight: bold; color: #e56922ff;">Travail à faire</span> :<br>
Écris un programme en Python qui extrait toutes les notes supérieures à la moyenne et les stocke dans une nouvelle liste.


````python
# Challenge 1 : Écrivez votre code ici
notes = [12, 4, 14, 11, 18, 13, 7, 10, 5, 9, 15, 8, 14, 16]

````

<details>
<summary>💡 <b>Cliquer ici pour afficher la solution du Challenge 1</b></summary>

```python
notes = [12, 4, 14, 11, 18, 13, 7, 10, 5, 9, 15, 8, 14, 16]
notes2 = []

# Calcul de la moyenne
some = 0
for note in notes:
    some += note
moyenne = some / len(notes)

# Extraction des notes supérieures à la moyenne
for note in notes:
    if note > moyenne:
        notes2.append(note)

print(f"Moyenne : {moyenne:.2f}")
print("Notes supérieures à la moyenne :", notes2)

# Alternative pythonique :
# moyenne = sum(notes) / len(notes)
# notes2 = [n for n in notes if n > moyenne]
```

<center><code>

</code></center>

<span style="font-weight: bold; color: #e56922ff;">Travail à faire</span> :<br>
Écris un programme en Python qui permet de regrouper dans une liste les mots communs entre deux chaînes de caractères Ch1 et Ch2.


````python
# Challenge 2 : Écrivez votre code ici
Ch1 = "Le langage Python est très populaire"
Ch2 = "Python est un langage puissant"

````

<details>
<summary>💡 <b>Cliquer ici pour afficher la solution du Challenge 2</b></summary>

```python
Ch1 = "Le langage Python est très populaire"
Ch2 = "Python est un langage puissant"

ch1_words = Ch1.split()
ch2_words = Ch2.split()
common_words = set()

for w1 in ch1_words:
    for w2 in ch2_words:
        if w1 == w2:
            common_words.add(w1)

# Conversion en liste demandée
resultat = list(common_words)
print("Mots communs :", resultat)

# Alternative pythonique avec intersection de sets :
# resultat = list(set(Ch1.split()) & set(Ch2.split()))
```

Écris un programme Python qui effectue les opérations suivantes à partir de la liste stock :

1. Afficher la liste initiale.
2. Créer deux nouvelles listes :
   * une contenant uniquement les chaînes de caractères,
   * l’autre contenant uniquement les valeurs numériques.
3. Trier :
   * la liste des chaînes en ordre croissant (alphabétique),
   * la liste des nombres en ordre décroissant.
4. Afficher les deux listes triées.


````python
# Challenge 3 : Écrivez votre code ici
stock = ["Stylo", 25, "Classeur", 100, "Crayon", 12, "Surligneur", 40, "Feutre", 5]

````

<details>
<summary>💡 <b>Cliquer ici pour afficher la solution du Challenge 3</b></summary>

```python
stock = ["Stylo", 25, "Classeur", 100, "Crayon", 12, "Surligneur", 40, "Feutre", 5]

# 1. Afficher la liste initiale
print("1. Liste initiale :", stock)

# 2. Séparer chaînes et nombres
strlist = []
numlist = []

for item in stock:
    if isinstance(item, int):
        numlist.append(item)
    elif isinstance(item, str):
        strlist.append(item)

# 3. Trier
strlist.sort()                # ordre croissant (alphabétique)
numlist.sort(reverse=True)    # ordre décroissant

# 4. Afficher les deux listes triées
print("2. Chaînes triées (A-Z) :", strlist)
print("3. Nombres triés (décroissant) :", numlist)
```

````python
Fruits = ["pomme", "banane", "orange", "kiwi", "mangue"]
````

```python
        if str(liste[i]).lower() == str(element).lower():

    return False

# Tests :
```

```python
L = [7, 23, 5, 23, 7, 19, 23, 12, 29]

def counterOfOccurrences(L, a):
    count = 0
    for item in L:
        if item == a:
            count += 1
    return count

# Test :
elem = 23
occ = counterOfOccurrences(L, elem)
print(f"L'élément {elem} est répété {occ} fois")
```
